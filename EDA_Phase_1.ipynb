```python
import pandas as pd

# Load dataset
df = pd.read_csv("player_summary_final.csv")

# 1. Check dataset shape
print("Dataset Shape:")
print(df.shape)

# 2. Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Check data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Check missing values in each column

print("Missing Values:")
print(df.isnull().sum())
print("Total Missing Values:", df.isnull().sum().sum())


```

```python
# Check duplicate records

print("Total Duplicate Rows:", df.duplicated().sum())

# Display duplicate records

duplicates = df[df.duplicated(keep=False)]

print(duplicates)
```

```python
# Count zero values in each column

zero_counts = (df == 0).sum()

print("Zero Values in Each Column:")
print(zero_counts)


# Total zero values in the dataset

print("Total Zero Values:", (df == 0).sum().sum())
```

```python
# Create batting and bowling indicators

df['is_batting'] = df['bat_balls'] >= 30

df['is_bowling'] = df['bowl_balls'] >= 30

# Display the results

print(df[['player', 'phase', 'bat_balls',
          'is_batting', 'bowl_balls',
          'is_bowling']].head(20))

print("Eligible batting records:",
      df['is_batting'].sum())

print("Eligible bowling records:",
      df['is_bowling'].sum())
```

```python
# Keep only the three required phases

df = df[df['phase'].isin(['Opening', 'Mid', 'Death'])].copy()

# Create separate batting dataset

batsmen_df = df[df['bat_balls'] >= 30].copy()

# Create separate bowling dataset

bowlers_df = df[df['bowl_balls'] >= 30].copy()

# Display results

print("Batsmen Dataset Shape:", batsmen_df.shape)

print("Bowlers Dataset Shape:", bowlers_df.shape)

print("BATSMEN - Records per Phase")
print(batsmen_df['phase'].value_counts())

print("\nBOWLERS - Records per Phase")
print(bowlers_df['phase'].value_counts())

# Batting table

batting_table = batsmen_df[
    ['player', 'phase', 'matches', 'bat_balls',
     'bat_runs', 'bat_avg', 'bat_sr']
]

print("BATSMEN TABLE")
display(batting_table.head(10))


# Bowling table

bowling_table = bowlers_df[
    ['player', 'phase', 'matches', 'bowl_balls',
     'bowl_wkts', 'bowl_avg', 'bowl_economy']
]

print("BOWLERS TABLE")
display(bowling_table.head(10))
```

```python
# Check duplicate player-phase combinations

print("Batsmen duplicate records:")
print(batsmen_df.duplicated(
    subset=['player', 'phase']
).sum())

print("\nBowlers duplicate records:")
print(bowlers_df.duplicated(
    subset=['player', 'phase']
).sum())

print("Batsmen phases:", batsmen_df['phase'].unique())

print("Bowlers phases:", bowlers_df['phase'].unique())
```

```python
# Save batting dataset
batsmen_df.to_csv(
    'batsmen_eda.csv',
    index=False
)

# Save bowling dataset
bowlers_df.to_csv(
    'bowlers_eda.csv',
    index=False
)

print("Both datasets saved successfully!")
```
