```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualisation style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Load batting dataset
batsmen_df = pd.read_csv('batsmen_eda.csv')

# Load bowling dataset
bowlers_df = pd.read_csv('bowlers_eda.csv')

print("Batsmen Dataset:", batsmen_df.shape)
print("Bowlers Dataset:", bowlers_df.shape)
```

```python
display(batsmen_df.head())

display(bowlers_df.head())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=batsmen_df,
    x='bat_runs',
    bins=20,
    color='skyblue'
)

plt.title('Distribution of Batting Runs')
plt.xlabel('Runs Scored')
plt.ylabel('Frequency')

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=batsmen_df,
    x='bat_runs',
    fill=True,
    color='orange',
    cut=0,
    clip=(0, None)
)

plt.title('Density Plot of Batting Runs')
plt.xlabel('Runs Scored')
plt.ylabel('Density')

plt.show()
```

```python

plt.figure(figsize=(8, 4))

sns.boxplot(
    data=batsmen_df,
    x='bat_runs',
    color='skyblue'
)

plt.title('Boxplot of Batting Runs')
plt.xlabel('Runs Scored')

plt.show()
```

```python

Q1 = batsmen_df['bat_runs'].quantile(0.25)
Q3 = batsmen_df['bat_runs'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = batsmen_df[
    (batsmen_df['bat_runs'] < lower_limit) |
    (batsmen_df['bat_runs'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of outliers:", len(outliers))

display(outliers[['player', 'bat_runs']])
```
